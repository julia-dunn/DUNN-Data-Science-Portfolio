# Part 1: Data Loading and Exploration

```python
import pandas as pd
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()

X = pd.DataFrame(housing.data, columns=housing.feature_names)
y = pd.Series(housing.target, name = 'med_house_value')
```

```python
print(X.head())
print(y.head())
```

```python
print(f'Features: {housing.feature_names}')
```

```python
X.info() # there are no missing values, all non-null counts = total number of entries
```

```python
X.describe() # basic way to describe data
```

```python
# method from week 7 notebook
scales = X.describe().T[["mean", "std", "min", "max"]]
scales["range"] = scales["max"] - scales["min"]

print("Feature scales, in their original units:")
scales.round(2)
```

# Part 2: Linear Regression on Uscaled Data

```python
# code from week 7 notebook
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, root_mean_squared_error, r2_score

# SPLIT FIRST - this is always step one
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=25)

print(f"Training set: {X_train.shape[0]:>5} rows")
print(f"Test set:     {X_test.shape[0]:>5} rows")
```

```python
# code from week 7 notebook
# Train on the RAW (unscaled) features
lin_reg_raw = LinearRegression()
lin_reg_raw.fit(X_train, y_train)

# Predict on the raw test data
y_pred_raw = lin_reg_raw.predict(X_test)

# Evaluate
mse_raw  = mean_squared_error(y_test, y_pred_raw)
rmse_raw = root_mean_squared_error(y_test, y_pred_raw)
r2_raw   = r2_score(y_test, y_pred_raw)

print("Raw (Unscaled) Data Model:")
print(f"Mean Squared Error:      {mse_raw:.4f}")
print(f"Root Mean Squared Error: {rmse_raw:.4f}")
print(f"R2 Score:                {r2_raw:.4f}")
print("\nModel Coefficients (Raw):")
print(pd.Series(lin_reg_raw.coef_, index = X_train.columns))
print(f"\nIntercept: {lin_reg_raw.intercept_:.4f}")
```

```python
# predictions

comparison_df = pd.DataFrame({
    'Actual': y_test,
    'Predicted': y_pred_raw,
    'Residual': y_test - y_pred_raw
})

comparison_df.head(10)
```

# Part 3: Feature Selection and Simplified Model

```python
X_2 = X[['AveBedrms', 'MedInc','Longitude']]
y_2 = y
```

```python
X_2_train, X_2_test, y_train, y_test = train_test_split(X_2, y, test_size=0.2, random_state=25)

print(f"Training set: {X_2_train.shape[0]:>5} rows")
print(f"Test set:     {X_2_test.shape[0]:>5} rows")
```

```python
# code from week 7 notebook
# Train on the RAW (unscaled) features
lin_reg_2_raw = LinearRegression()
lin_reg_2_raw.fit(X_2_train, y_train)

# Predict on the raw test data
y_pred_2_raw = lin_reg_2_raw.predict(X_2_test)

# Evaluate
mse_2_raw  = mean_squared_error(y_test, y_pred_2_raw)
rmse_2_raw = root_mean_squared_error(y_test, y_pred_2_raw)
r2_2_raw   = r2_score(y_test, y_pred_2_raw)

print("Raw (Unscaled) Data Model:")
print(f"Mean Squared Error for Simplified Reg:      {mse_raw:.4f}")
print(f"Root Mean Squared Error for Simplified Reg: {rmse_raw:.4f}")
print(f"R2 Score for Simplified Reg:                {r2_raw:.4f}")
print("\nModel Coefficients (Raw) for Simplified Reg:")
print(pd.Series(lin_reg_2_raw.coef_, index = X_2_train.columns))
print(f"\nIntercept: {lin_reg_2_raw.intercept_:.4f}")
```

```python
# predictions
import numpy as np
comparison_df = pd.DataFrame({
    'Actual': y_test,
    'Predicted (original)': y_pred_raw,
    'Predicted (simplified)': y_pred_2_raw,
    'Original Residual': y_test - y_pred_raw,
    'Simplified Residual': y_test - y_pred_2_raw
})

comparison_df['Smaller Residual'] = np.where(
    comparison_df['Original Residual'].abs() < comparison_df['Simplified Residual'].abs(),
    'Original',
    'Simplified'
)

comparison_df.head(10)
```
